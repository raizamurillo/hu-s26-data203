```python
passage = """Ella Fitzgerald is an American jazz singer widely regarded as one of the greatest and most influential vocalists in
the history of jazz. Born on April 25, 1916, in Newport News, Virginia, she developed a distinctive voice and style
that emphasized precision, phrasing, and emotional expression.

Fitzgerald was known for her impeccable timing, dynamic range, and ability to improvise with grace and subtlety.
She was a master of scat singing and had a profound influence on the development of jazz vocal styles. Her work
with the legendary Louis Armstrong and her collaborations with musicians like Miles Davis, Duke Ellington, and
Ray Charles helped shape the sound of jazz music.

She was also a key figure in the “crossover” movement, bringing jazz to broader audiences through pop and film
music. Fitzgerald recorded over 1,000 songs and was inducted into the Grammy Hall of Fame and the National
Recording Registry.

Ella Fitzgerald passed away on June 15, 1996, in New York City, leaving behind a legacy that continues to inspire
generations of musicians and fans alike.
"""
```

```python
num_paragraphs = len(passage.splitlines())
print("Number of paragraphs:", num_paragraphs)
```

```python
passage_lower = passage.lower()
print(passage_lower)
```

```python
str.lower(passage)
passage_replaced = passage_lower.replace(" her ", " ### " )
print(passage_replaced)
```

```python
and_count = passage_lower.count(" and ")
print("NUmber of times 'and' appears:", and_count)
```

```python
jazz_count = passage_lower.count("jazz")
print("Number of times 'jazz' appears:", jazz_count)
```

```python
paragraphs = passage.split("\n\n")
print(paragraphs)
```

```python
first_paragraph = paragraphs[0]
print(first_paragraph)
```

```python
words = passage.split()
ing_words = [word.strip() for word in words if word.strip().lower().endswith("ing")]
print(f"number of words ending in 'ing': {len(ing_words)}\n")
print("words ending in 'ing':")
for word in ing_words:
  print(f"- {word}")
```

```python
er_words = [word.strip() for word in words if word.strip().lower().endswith("er")]
print(f"Number of words ending in 'er': {len(er_words)}\n")
print("Words ending in 'er':")
for word in er_words:
    print(f"- {word}")
```
